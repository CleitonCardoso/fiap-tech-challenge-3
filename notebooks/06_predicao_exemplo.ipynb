# 06 - Predição com modelo exportado

Exemplo de como carregar o bundle salvo em `models/random_forest_delay_model.pkl` e gerar previsões aplicando o threshold calibrado.

```python
# Carrega modelo + threshold do arquivo salvo
import joblib
import pandas as pd
from pathlib import Path

bundle_path = Path("../models/random_forest_delay_model.pkl")
bundle = joblib.load(bundle_path)
model = bundle["model"]
threshold = bundle["threshold"]

features = [
    "AIRLINE",
    "ORIGIN_AIRPORT",
    "DESTINATION_AIRPORT",
    "MONTH",
    "DAY_OF_WEEK",
    "DEP_HOUR",
    "DISTANCE",
    "IS_WEEKEND",
]
```

```python
# Exemplo de entrada: um voo AA de JFK para LAX numa sexta (dia 5) de julho, decolando 14h
X_new = pd.DataFrame([
    {
        "AIRLINE": "AA",
        "ORIGIN_AIRPORT": "JFK",
        "DESTINATION_AIRPORT": "LAX",
        "MONTH": 7,
        "DAY_OF_WEEK": 5,  # 1=segunda, ... 7=domingo, conforme dataset
        "DEP_HOUR": 14,
        "DISTANCE": 3983,
        "IS_WEEKEND": 0,
    }
])

# Garante que as colunas estejam na ordem esperada
X_new = X_new[features]
```

```python
# Probabilidade de atraso e classe usando o threshold calibrado
probas = model.predict_proba(X_new)[:, 1]
preds = (probas >= threshold).astype(int)

print(f"Threshold salvo: {threshold:.3f}")
print(f"Probabilidade de atraso: {probas[0]:.3f}")
print(f"Classe prevista (1=atraso): {preds[0]}")
```
